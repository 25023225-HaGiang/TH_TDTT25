```python
#W7A16:
a=list(map(int, input().split()))
a.sort()
cnt={}
for i in range(len(a)):
    if a[i] not in cnt:
        cnt[a[i]]=1
    else:
        cnt[a[i]]+=1
ktra=True
them=0
for k in cnt:
    if cnt[k]<k:
        them+=cnt[k]
        ktra=False
    elif cnt[k]>k:
        them+=cnt[k]-k
        ktra=False
if ktra:
    print("YES")
else:
    print("NO")
    print(them)
```

```python
#W7A17
a=list(map(int,input().split()))
cand=0
cnt=0
for x in a:
	if cnt==0:
		cand=x;cnt=1
	elif x==cand:
		cnt+=1
	else:
		cnt-=1
if a.count(cand)>len(a)//2:
	print(cand)
    
```

```python
#W7A19
a=list(map(int,input().split()))
b=list(map(int,input().split()))
i=j=0
res=[]
inc=a[0]<=a[-1]
while i<len(a) and j<len(b):
	if (a[i]<=b[j])==inc:
		res.append(a[i]);i+=1
	else:
		res.append(b[j]);j+=1
res+=a[i:]+b[j:]
print(*res)
```

```python
#W7A25
s=input().strip()
k=int(input())
st=[]
for i,x in enumerate(s):
	while st and k>0 and st[-1]<x:
		st.pop();k-=1
	st.append(x)
print("".join(st[:len(s)-k]))
```

```python
#W7A27
s=input()
st=[]
ok=True
mp={')':'(','}':'{',']':'[','>':'<'}
for c in s:
	if c in '({[<':
		st.append(c)
	else:
		if not st or st.pop()!=mp.get(c,''):
			ok=False;break
print("YES" if ok and not st else "NO")
```
