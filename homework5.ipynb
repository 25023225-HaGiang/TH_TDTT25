```python
#W6A16:Nhâp vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma 
#trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải
n, m = map(int, input().split())
a=[]
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)
for i in range(n):
    for j in range(m):
        print(f"{a[i][j]:>4}", end="")
    print()
```

```python
#W6A18:Nhâp vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
#cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
#k(hàng k)
n, m, k = map(int, input().split())
a = []
for i in range(n):
    row = list(map(int, input().split()))
    a.append(row)
tongcotk = 0
for i in range(n):
    tongcotk += a[i][k-1]
tonghangk = 0
for i in range(m):
    tonghangk += a[k][i]
print("Tổng cột k=", tongcotk)
print("Tổng hàng k=", tonghangk)
```

```python
#W6A19:Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
#trong list
lst = list(map(int, input().split()))
def pos_max(lst):
    pmax = lst[0]
    pos = 0
    for i in range(1, len(lst)):
        if lst[i] > pmax:
            pmax = lst[i]
            pos = i + 1
    return pos
print(pos_max(lst))
```

```python
#W6A20:Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
#trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1
lst = list(map(int, input().split()))
k = int(input())
pos = 0
for i in range( len(lst)):
    if lst[i] == k:
        pos = i +1
        break
    else: pos = -1
print(pos)
        
    
```

```python
#W6A21: Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
#chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện
lst = list(input().split())
d = {}
for i in range(len(lst)):
    word = lst[i]
    if word not in d:
        d[word] = []
    d[word].append(i + 1)
for j in d:
    d[j] = tuple(d[j]) 
print(d)
    
```

```python
#W6A22: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
#trong list
lst = list(input().split())
k = max(lst)
for i in range (len(lst)):
    if k == lst[i]:
        print(i+1)
        break
```

```python
#W6A23:Viết hàm xóa đi các số <x có trong mảng
lst = list(map(int, input().split()))
k = int(input())
nlst = []
for i in range (len(lst)):
    if lst[i] >= k:
        nlst.append(lst[i])
print(nlst)
```

```python
#W6A24: Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
#trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9
lst = list(map(int, input().split()))
zr = []
other = []
for i in range (len(lst)):
    if lst[i] == 0:
        zr.append(lst[i])
    if lst[i] != 0:
        other.append(lst[i])
print( zr + other)
        
```

```python
#6A26:Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
#phần tử đầu và cuối
lst = list(map(int, input().split()))
tong = 0
for i in range (1, len(lst) - 1):
    if lst[i] > lst[i-1] and lst[i] > lst[i+1]:
        tong += lst[i]
print(tong)
```

```python
#6A27:Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
#nếu không có in ra -1
lst = list(map(int, input().split()))
k = int(input())
nlst = []
for i in range (len(lst)):
    if lst[i] < k:
        nlst.append(i+1)
if len(nlst) == 0:
    print("-1")
else:
    print(*nlst)
    
```

```python
#W6A28:Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
#thực sự, hay vô danh (không tăng, không giảm)
lst = list(map(int, input().split()))
inc = True
dec = True
for i in range (len(lst)-1):
    if lst[i] > lst[i+1]:
        inc = False
    if lst[i] < lst[i+1]:
        dec = False
if inc == True:
    print("La day tang that su")
elif dec == True:
    print("La day giam that su")
else: 
    print("La day vo danh")
```
