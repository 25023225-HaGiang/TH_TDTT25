```python
#BT10
x, y  = map(float, input("Nhập tọa độ tâm của đường tròn x, y:").split())
r = float(input("Nhập vào bán kính của đường tròn:"))
x1, y1 = map(float, input("Nhập tọa độ điểm A(x,y):").split())
ptbk = (x-x1)**2 + (y-y1)**2
if ptbk == r**2:
    print("Điểm A thuộc đường tròn")
elif ptbk < r**2:
    print("Điểm A nằm trong đường tròn")
else: print("Điểm A nằm ngoài đường tròn")
          
```
