```python
#W5A1:Nhập vào hai số a,b in ra số lớn hơn trong hai số
def max_of_two(a:int, b:int):
    return max(a, b)
try:
    a, b = map(int, input().split())
    print(max_of_two(a, b))
except ValueError:
    print("Nhap du lieu loi")
```

```python
#5A2:Nhập vào 2 số a,b in ra b,a
a, b = map(int, input().split())
def swap(a, b):
    a,b = b,a
    return a,b
a, b = swap(a,b)
print(a,b)
```

```python
#5A3:Nhập vào số n, in ra các số nguyên tố từ 1 đến n
import math
def so(n: int) -> bool:
    for i in range(2,math.isqrt(n)+1):
        if n % i == 0:
            return False
    return (n>=2)
n = int(input())
for x in range(2, n+1):
    if (so(x)):
        print(x, end=" ")
    
```

```python
#5A4:Viết hàm kiểm tra một số có phải số hoàn hảo không, xử lý dữ liệu kiểu số nguyên
def shh(n: int):
    tong = 0
    for i in range (1, n):
        if n % i == 0:
            tong += i
    if tong == n:
        return True
try: 
    n = int(input("Nhập vào 1 số nguyên:"))
    if (shh(n)):
        print("Đây là số hoàn hảo")
    else: print("Đây không phải số hoàn hảo")
except ValueError:
    print("Vui lòng nhập 1 số nguyên")
```

```python
#5A5:Viết đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên
#có giá trị k trong list số nguyên, nếu không có thì trả về -1
def vtri(lst, k):
    for i in range ( len(lst)):
        if lst[i] == k:
            return i + 1
    return -1
try:
    lst = list(map(int, input("Nhập 1 dãy số các số nguyên").split()))
    k = int(input("Nhập vào 1 số nguyên dương:"))
    if vtri(lst, k) != -1:
        print("Vị trí của số nguyên dương k trong dãy số là", vtri(lst, k))
    else: print("-1 (Số không có trong dãy")
except ValueError:
    print("vui lòng nhập đúng yêu câu!")
    

```
