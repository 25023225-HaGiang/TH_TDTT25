```python
#W4A2
import math 
n = int(input("Nhập 1 số nguyên dương n"))
if n >= 0:
    for i in range (2,math.isqrt(n)+1):
        uoc = n%i
        if uoc == 0:
            print("Đây không phải số nguyên tố")
            break
    else: print("Đây là số nguyên tố")
else: print("Nhập 1 số nguyên dương")
```

```python
#W4A6

import math
tong = 0
a, b = map(int, input("Nhập hai số nguyên dương").split())
if a > b:
    print("Nhập lại")
else:
    def songuyento(n):
        if n < 2:
            return False
        if n == 2:
            return True
        if n % 2 == 0:
            return False
        for i in range (3, math.isqrt(n)+1, 2):
            if n % i == 0:
                return False
        return True
    for i in range (a, b+1):
        if songuyento(i):
            tong += i
    print("Tổng của các só nguyên tố từ", a, "đến", b, "là", tong)
            
        
    
```

```python
#W4A7
import math
n = int(input("Nhập số nguyên dương n (n>=2)"))
if n > 2:
    def sont(p):
        for i in range (2, math.isqrt(p)+1, 1):
            if p % i == 0:
                return False
        return True
    umax = 1
    for x in range (1, n+1):
        if n % x == 0:
            if sont(x):
                umax = x
    print("Ước là số nguyên tố lớn nhất của", n,"là", umax)
elif n == 2:
    print("Ước nguyên tố lớn nhất là 2")
else: print("Vui lòng nhập lại")
                
        
            
    
```
