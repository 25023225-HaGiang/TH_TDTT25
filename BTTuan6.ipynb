```python
#W6A1:
lst = list(map(int, input().split()))
S = []
for i in lst:
    if i not in S:
        S.append(i)
print(S)
```

```python
#W6A2:
lst = list(map(int, input().split()))
S = []
tong = 0
for i in lst:
    tong += i
    S.append(tong)
print(S)
```

```python
#W6A3:
lst = tuple(map(int, input().split()))
k = int(input())
kq = lst[k:]+lst[:k]
print(kq)
```

```python
#W6A4:
lst = list(input().split())
ax = {}
for i in lst:
    key, value = i.split(":")
    if key not in ax:
        ax[key] = []
    ax[key].append(value)
print(ax)
```

```python
#W6A5:
lst = list(map(float, input().split()))
kq = {'positives': 0,
      'negatives': 0,
      'zeros': 0}
for i in lst:
    if i == 0:
        kq['zeros'] += 1
    if i < 0:
        kq['negatives'] += 1
    if i > 0:
        kq['positives'] += 1
print(kq)
        
```
