```python
#W6A16:Nhâ ̣p vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma
#trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải
n, m = map(int, input().split())

lst = []
for i in range (n):
    lst.append(list(map(int, input().split())))
for i in range (n):
    for j in range (m):
        print(f"{lst[i][j]:>4}", end ="")
    print()
```

```python
#W6A17:Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và 
#đường chéo phụ của mảng
n = int(input())
lst = []
for i in range (n):
    lst.append(list(map(int, input().split())))
for i in range (n):
    print(lst[i][i], end =" ") 
```

```python
#W6A18:Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
#cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
#k (hàng k) 
n, m, k = list(map(int, input().split()))
lst = []
th = 0
tc = 0
for i in range (n):
    lst.append(list(map(int, input().split())))
for i in range (n):
        th += lst[k-1][i]
for i in range (m):
        tc += lst[i][k-1]
print(th, tc)
    
    
    
```
