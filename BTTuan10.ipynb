```python
#W9A1:
def merge(a, b):
    if not a:
        return b
    if not b:
        return a
    if a[0] < b[0]:
        return [a[0]] + merge(a[1:], b)
    else: return [b[0]]+ merge(b[1:], a)
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(*merge(a,b))
```

```python
#W9A3
def helper(s, i, best):
    if i == len(s):
        return best
    for j in range(i + 1, len(s)):
        lst = list(s)
        lst[i], lst[j] = lst[j], lst[i]
        new_s = "".join(lst)

        if new_s[0] != '0':   # tránh số bắt đầu bằng 0
           if best is None or new_s < best:
               best = new_s
    return helper(s, i + 1, best)
def min_n(n):
    return helper(n, 0, None)
n = input().strip()
print(min_n(n))
```
