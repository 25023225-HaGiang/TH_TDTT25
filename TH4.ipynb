```python
#W4A2

for _ in range(100):
    n = int(input("Nhập vào một số nguyên dương"))
    if(n>0):
        break
    else:
        print("Nhập 1 số nguyên dương!")
print("Số nguyên dương bạn nhập được là:", n)
cnt = 0 #dem so uoc
for i in range(2, int(n**(1/2)+1)):
    if (n%i == 0):
        cnt += 1 
if (cnt == 2):
    print(n, "là số nguyên tố")
else:
    print(n, "không phải là số nguyên tố")
```

```python
#4A4
try: 
    n = int(input())
    n = abs(n)
    count = 0
    if n==0:
        print(1)
    else:
        for _ in range(100):
            count += 1
            n //= 10
            if n == 0:
                break
            
        print(count)
except ValueError:
    print("Bạn đã nhập sai dữ liệu")
```

```python
#W4A13
try:
    a = int(input())
    b = int(input())
    tonguoca = 0
    for i in range(1, a//2 + 1):
        if a%i == 0:
            tonguoca += i 
    tonguocb= 0    
    for i in range(1, b//2 + 1):
        if b%i == 0:
            tonguocb += i 
    if tonguoca == b and tonguocb == a:
        print("True")
    else:
        print("False")
except ValueError:
    print("Bạn đã nhập sai")
```
