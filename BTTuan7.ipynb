```python
#W7A2:Cho danh sách arr và một số nguyên x, đếm xem x
#xuất hiện bao nhiêu lần trong danh sách. def
#count_occurrences(arr, x) -> int:
def count_occurrences(arr, x):
    cnt = 0
    for i in range (len(arr)):
        if arr[i] == x:
            cnt += 1
    return cnt

arr = list(map(int, input().split()))
x = int(input())
print(count_occurrences(arr, x))
```

```python
#W7A3:Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách
#theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
#Sort):
#Bubble sort:
lst = list(map(int, input().split()))
def bubble_sort(lst):
    for i in range (len(lst)):
        swap = False
        for j in range (0, len(lst)-i-1):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+i] = lst[j+i], lst[j]
                swap = True
        if not swap:
            break
    return lst
def selection_sort(lst):
    for i in range (len(lst)):
        min_idx = i
        for j in range(i+1, len(lst)):
            if lst[j] < lst[min_idx]:
                min_idx = j
        lst[i], lst[min_idx] = lst[min_idx], lst[i]
    return lst
lst = list(map(float, input().split()))
print(bubble_sort(lst))
print(selection_sort(lst))         
```

```python
#W7A4:Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong
#danh sách. Nếu có nhiều phần tử cùng số lần trả về phần tử xuất hiện trước.
arr = list(map(int, input().split()))
ts = [1]*len(arr)
for i in range(len(arr)):
    for j in range (i):
        if arr[i] == arr[j]:
            ts[i] += 1
m_ts = max(ts)
vtri = 1
for i in range(len(ts)):
    if ts[i] == m_ts:
        vtri = i
        break
print(m_ts, arr[vtri])
```

```python
#W7A10
def max_subarray_mod(nums,m):
 pre=0
 best=0
 s=[]
 import bisect
 for x in nums:
  pre=(pre+x)%m
  i=bisect.bisect_right(s,pre)
  if i<len(s):
      best=max(best,(pre-s[i]+m)%m)
  else:
      best=max(best,pre)
  bisect.insort(s,pre)
 return best
a=list(map(int,input().strip()[1:-1].split(",")))
m=int(input())
result=max_subarray_mod(a,m)
print(str(result))  
```
